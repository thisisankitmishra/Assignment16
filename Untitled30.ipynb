# ---------------------------------------------------------
# Question 1: What is Anomaly Detection? Types with examples
# ---------------------------------------------------------

# Anomaly Detection is the process of identifying data points
# that significantly deviate from normal data behavior.
# These abnormal points are called anomalies or outliers.

# Types of Anomalies:
# 1. Point Anomaly:
#    A single data point that is very different from others.
#    Example: A bank transaction of Rs 5,00,000 when normal is Rs 5,000.

# 2. Contextual Anomaly:
#    An anomaly depending on context such as time or location.
#    Example: High electricity usage at midnight.

# 3. Collective Anomaly:
#    A group of data points that together form an anomaly.
#    Example: Continuous server downtime over a period of time.


# ---------------------------------------------------------
# Question 2: Compare Isolation Forest, DBSCAN, LOF
# ---------------------------------------------------------

# Isolation Forest:
# - Randomly splits data to isolate anomalies
# - Fast and works well on large datasets
# - Best for high-dimensional data

# DBSCAN:
# - Density-based clustering algorithm
# - Detects anomalies as low-density points
# - Best for spatial and clustering problems

# Local Outlier Factor (LOF):
# - Measures local density deviation
# - Detects local anomalies
# - Best when data has varying density


# ---------------------------------------------------------
# Question 3: Components of Time Series
# ---------------------------------------------------------

# 1. Trend:
#    Long-term increase or decrease
#    Example: Yearly increase in sales

# 2. Seasonality:
#    Repeating patterns over fixed intervals
#    Example: Higher sales during festivals

# 3. Cyclic:
#    Business or economic cycles
#    Example: Recession and economic growth

# 4. Irregular (Noise):
#    Random fluctuations
#    Example: Power failure affecting production


# ---------------------------------------------------------
# Question 4: Stationarity in Time Series
# ---------------------------------------------------------

# A time series is stationary if:
# - Mean is constant over time
# - Variance is constant
# - No seasonality or trend

# Stationarity Tests:
# - Augmented Dickey-Fuller (ADF) Test
# - KPSS Test

# Transformations to make series stationary:
# - Differencing
# - Log transformation
# - Seasonal differencing


# ---------------------------------------------------------
# Question 5: AR vs MA vs ARIMA vs SARIMA vs SARIMAX
# ---------------------------------------------------------

# AR (AutoRegressive):
# Uses past values

# MA (Moving Average):
# Uses past errors

# ARIMA:
# AR + MA + Differencing

# SARIMA:
# ARIMA + Seasonal components

# SARIMAX:
# SARIMA + Exogenous (external) variables

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

data = pd.read_csv("AirPassengers.csv")

# ---------------------------------------------------------
# Question 7: Isolation Forest on NYC Taxi Fare Dataset
# ---------------------------------------------------------

from sklearn.ensemble import IsolationForest

taxi = pd.read_csv("nyc_taxi.csv")

X = taxi[["fare_amount", "trip_distance"]]

iso_forest = IsolationForest(
    contamination=0.02,
    random_state=42
)

taxi["anomaly"] = iso_forest.fit_predict(X)

plt.scatter(
    X["fare_amount"],
    X["trip_distance"],
    c=taxi["anomaly"]
)
plt.xlabel("Fare Amount")
plt.ylabel("Trip Distance")
plt.show()


# ---------------------------------------------------------
# Question 8: SARIMA Forecasting (Next 12 Months)
# ---------------------------------------------------------

from statsmodels.tsa.statespace.sarimax import SARIMAX

sarima_model = SARIMAX(
    data["#Passengers"],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12)
)

sarima_result = sarima_model.fit()

forecast = sarima_result.get_forecast(steps=12)

plt.plot(data["#Passengers"], label="Actual")
plt.plot(forecast.predicted_mean, label="Forecast")
plt.legend()
plt.show()

from sklearn.neighbors import LocalOutlierFactor
